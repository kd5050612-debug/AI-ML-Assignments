```python
import matplotlib.pyplot as plt
```

```python


x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, color="blue", marker="o")

plt.title("Simple Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 30, 35]
y2 = [5, 15, 22, 56, 22]

plt.plot(x, y1, label="Dataset1")
plt.plot(x, y2, label="Dataset2")

plt.xlabel("x")
plt.ylabel("Value")

plt.title("Comparison of Datasets")

plt.legend()
plt.grid(True)
plt.show()
```

```python
marks = [35,42,50,55,65,72,78,80,85,50,92]
plt.hist(marks, bins=3)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

```python
plt.subplot(1,2,2)
plt.plot(x, y1)
plt.title("Line Plot")

plt.subplot(1,2,1)
plt.scatter(x,y1)
plt.title("Scatter Plot")

plt.show()
```

```python
subjects=["Python","ML","DBMS","AI"]
marks= [30, 25, 20, 25]
plt.pie(marks, labels=subjects, autopct="%0.1f%%")
plt.title("Marks Distribution")
plt.show()
```
