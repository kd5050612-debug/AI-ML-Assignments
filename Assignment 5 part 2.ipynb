```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Iris.csv")

print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.info())
```

```python
print(df.describe())
print(df.mean(numeric_only=True))
print(df.median(numeric_only=True))
print(df.std(numeric_only=True))
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))
```

```python
print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].mean())

print(df.isnull().sum())

print(df.duplicated().sum())

df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
print(df["Species"].unique())
print(df["Species"].nunique())
print(df["Species"].value_counts())
```

```python
print(df[df["SepalLengthCm"] > 5])

print(df.sort_values("PetalLengthCm").head(10))

print(
    df.groupby("Species")[
        ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
    ].mean()
)
```

```python
sepal_length = df["SepalLengthCm"].to_numpy()
```

```python
print(np.mean(sepal_length))
print(np.median(sepal_length))
print(np.std(sepal_length))
print(np.min(sepal_length))
print(np.max(sepal_length))
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
df.hist(figsize=(10, 8), bins=15)
plt.suptitle("Histograms")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=numeric_df)
plt.title("Box Plot")
plt.xticks(rotation=20)
plt.show()
```

```python
for column in numeric_df.columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]
    print(column, len(outliers))
```

```python

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
plt.title("Sepal Length vs Petal Length")
plt.show()

sns.pairplot(df, hue="Species")
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df,
    x="Species",
    y="PetalLengthCm"
)
plt.title("Petal Length by Species")
plt.show()

print(df.head())
print(df.shape)
df.info()
```
